# Supply Chain Risk Analysis
Combines sales forecasts, inventory, and supplier risks to identify supply chain vulnerabilities.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sales = pd.read_csv('../dataset/synthetic/sales.csv')
products = pd.read_csv('../dataset/synthetic/products.csv')
suppliers = pd.read_csv('../dataset/synthetic/suppliers.csv')
```

```python
daily_sales = sales.groupby('date')['daily_sales'].sum().reset_index()
daily_sales['date'] = pd.to_datetime(daily_sales['date'])

plt.figure(figsize=(14,6))
plt.plot(daily_sales['date'], daily_sales['daily_sales'])
plt.title('Overall Daily Sales Volume')
plt.show()
```
