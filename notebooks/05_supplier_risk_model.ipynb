# Supplier Risk Prediction (XGBoost + Optuna + SHAP)
Uses Optuna for hyperparameter tuning an XGBoost model that predicts supplier shipment delays.

```python
import polars as pl
import xgboost as xgb
import optuna
import shap
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error
```

```python
print('Loading shipments data...')
df = pl.read_csv('../dataset/synthetic/shipments.csv')
# Calculate delay
df = df.with_columns([
    pl.col('expected_delivery').str.to_datetime('%Y-%m-%d', strict=False).cast(pl.Date),
    pl.col('actual_delivery').str.to_datetime('%Y-%m-%d', strict=False).cast(pl.Date)
])
df = df.with_columns((pl.col('actual_delivery') - pl.col('expected_delivery')).dt.total_days().alias('delay_days'))
df = df.filter(pl.col('delay_days').is_not_null())
df.head()
```
