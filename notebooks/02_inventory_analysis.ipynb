# Inventory & Stockout Analysis
Analyzes product stock levels, safety stocks, and reorder points.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../dataset/synthetic/products.csv')
df.head()
```

```python
df['stock_status'] = 'Healthy'
df.loc[df['current_stock'] <= df['reorder_point'], 'stock_status'] = 'Reorder Needed'
df.loc[df['current_stock'] <= df['safety_stock'], 'stock_status'] = 'Critical'

status_counts = df['stock_status'].value_counts()
status_counts.plot(kind='pie', autopct='%1.1f%%', title='Inventory Status Distribution')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df['current_stock'], bins=30, kde=True)
plt.title('Distribution of Current Stock Levels')
plt.show()
```
